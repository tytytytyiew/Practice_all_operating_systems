```python
%%writefile benchmark.py
import time
import statistics
import requests
from concurrent.futures import ThreadPoolExecutor, ProcessPoolExecutor

WORKERS = 4
REPEATS = 3

with open("numbers.txt") as f:
    numbers = [int(line) for line in f]

def sum_chunk(chunk):
    return sum(chunk)

def chunks_of_four(data):
    size = (len(data) + WORKERS - 1) // WORKERS
    return [data[i:i + size] for i in range(0, len(data), size)]

def benchmark_sum(method):
    times = []
    chunks = chunks_of_four(numbers)

    for _ in range(REPEATS):
        start = time.perf_counter()

        if method == "Один поток":
            result = sum(numbers)
        elif method == "4 потока":
            with ThreadPoolExecutor(max_workers=WORKERS) as pool:
                result = sum(pool.map(sum_chunk, chunks))
        else:
            with ProcessPoolExecutor(max_workers=WORKERS) as pool:
                result = sum(pool.map(sum_chunk, chunks))

        times.append(time.perf_counter() - start)

    return result, statistics.median(times)

print("=== СУММИРОВАНИЕ ЧИСЕЛ ===")
print(f"Количество чисел: {len(numbers):,}")
print(f"{'Способ':<20} {'Сумма':>15} {'Медиана, сек.':>15}")

for method in ["Один поток", "4 потока", "4 процесса"]:
    result, median = benchmark_sum(method)
    print(f"{method:<20} {result:>15,} {median:>15.4f}")

URLS = [
    "https://example.com",
    "https://www.python.org",
    "https://www.wikipedia.org",
    "https://www.github.com",
    "https://www.google.com",
    "https://www.mozilla.org",
    "https://www.microsoft.com",
    "https://www.apple.com",
    "https://www.bing.com",
    "https://www.cloudflare.com",
    "https://www.djangoproject.com",
    "https://pypi.org",
    "https://docs.python.org",
    "https://www.gnu.org",
    "https://www.kernel.org",
    "https://www.nasa.gov",
    "https://www.un.org",
    "https://www.who.int",
    "https://www.kaggle.com",
    "https://httpbin.org/get",
]

def download(url):
    try:
        response = requests.get(url, timeout=8)
        return len(response.content), response.status_code
    except requests.RequestException:
        return 0, "ERROR"

def benchmark_network(method):
    times = []

    for _ in range(REPEATS):
        start = time.perf_counter()

        if method == "Последовательно":
            results = [download(url) for url in URLS]
        elif method == "ThreadPoolExecutor":
            with ThreadPoolExecutor(max_workers=WORKERS) as pool:
                results = list(pool.map(download, URLS))
        else:
            with ProcessPoolExecutor(max_workers=WORKERS) as pool:
                results = list(pool.map(download, URLS))

        times.append(time.perf_counter() - start)

    success = sum(
        1 for size, status in results
        if status != "ERROR" and 200 <= status < 400
    )

    return statistics.median(times), success

print("\n=== ЗАГРУЗКА 20 URL ===")
print(f"{'Способ':<25} {'Медиана, сек.':>15} {'Успешно':>12}")

for method in [
    "Последовательно",
    "ThreadPoolExecutor",
    "ProcessPoolExecutor",
]:
    median, success = benchmark_network(method)
    print(f"{method:<25} {median:>15.3f} {success:>9}/20")
```

```python
%%writefile memory_demo.py
import threading
import multiprocessing

data = [10, 20, 30]

def change_data():
    data[0] = 999
    print("Внутри потока:", data)

def change_process_data():
    data[0] = 555
    print("Внутри процесса:", data, flush=True)

if __name__ == "__main__":
    print("Исходный список:", data)

    thread = threading.Thread(target=change_data)
    thread.start()
    thread.join()

    print("После потока:", data)

    data[0] = 10

    process = multiprocessing.Process(target=change_process_data)
    process.start()
    process.join()

    print("После процесса:", data)
```
