```python
%%bash
cat > lab/report.sh <<'EOF'
#!/usr/bin/env bash

usage() {
    echo "Использование: $0 <каталог> <ERROR|WARN> [--top N]"
    echo "Пример: $0 lab/logs ERROR --top 3"
}

if [ "$#" -eq 0 ]; then
    usage
    exit 1
fi

if [ "$#" -lt 2 ]; then
    usage
    exit 1
fi

DIR="$1"
LEVEL="$2"
TOP=0

if [ ! -d "$DIR" ]; then
    echo "Ошибка: каталог '$DIR' не существует." >&2
    exit 1
fi

if [[ "$LEVEL" != "ERROR" && "$LEVEL" != "WARN" ]]; then
    echo "Ошибка: уровень должен быть ERROR или WARN." >&2
    usage
    exit 1
fi

if [ "$#" -gt 2 ]; then
    if [ "$#" -ne 4 ] || [ "$3" != "--top" ]; then
        usage
        exit 1
    fi

    if ! [[ "$4" =~ ^[1-9][0-9]*$ ]]; then
        echo "Ошибка: N должно быть положительным целым числом." >&2
        exit 1
    fi

    TOP="$4"
fi

if ! find "$DIR" -type f -print -quit | grep -q .; then
    echo "Ошибка: каталог не содержит файлов."
    exit 1
fi

echo "Модуль → число сообщений ($LEVEL)"
printf '%s\n' "--------------------------------"

RESULT=$(find "$DIR" -type f -exec awk -v level="$LEVEL" '
    $1 == level { count[$2]++ }
    END {
        for (module in count)
            print count[module], module
    }
' {} + | sort -k1,1nr -k2,2)

if [ -z "$RESULT" ]; then
    echo "Сообщения уровня $LEVEL не найдены."
    exit 0
fi

if [ "$TOP" -gt 0 ]; then
    RESULT=$(printf '%s\n' "$RESULT" | head -n "$TOP")
fi

while read -r COUNT MODULE; do
    printf '%-15s → %s\n' "$MODULE" "$COUNT"
done <<< "$RESULT"
EOF

chmod +x lab/report.sh
echo "report.sh создан"
```

```python
%%bash
cat > lab/hello.py <<'EOF'
import sys
from pathlib import Path

if len(sys.argv) != 2:
    print(f"Использование: python3 {sys.argv[0]} <файл>")
    sys.exit(1)

file_path = Path(sys.argv[1])

try:
    data = file_path.read_bytes()
    print(f"Файл: {file_path}")
    print(f"Размер: {len(data)} байт")
except OSError as error:
    print(f"Ошибка чтения файла: {error}", file=sys.stderr)
    sys.exit(1)
EOF

echo "hello.py создан"
```

```python
%%bash
cd lab

CAT_COUNT=$(grep -Ec '^[[:space:]]*([0-9]+[[:space:]]+)?[[:alpha:]_][[:alnum:]_]*\(' cat_trace.txt)
PYTHON_COUNT=$(grep -Ec '^[[:space:]]*([0-9]+[[:space:]]+)?[[:alpha:]_][[:alnum:]_]*\(' python_trace.txt)

echo "Системные вызовы cat: $CAT_COUNT"
echo "Системные вызовы Python: $PYTHON_COUNT"

if [ "$PYTHON_COUNT" -gt "$CAT_COUNT" ]; then
    echo "Python выполнил больше системных вызовов."
elif [ "$PYTHON_COUNT" -lt "$CAT_COUNT" ]; then
    echo "cat выполнил больше системных вызовов."
else
    echo "Количество системных вызовов одинаковое."
fi
```

```python
%%bash
cd lab

TOTAL=$(grep -Ec '(^|[[:space:]])[[:alpha:]_][[:alnum:]_]*\(' trace.txt)
READ_COUNT=$(grep -Ec '(^|[[:space:]])read\(' trace.txt)
CAT_COUNT=$(grep -Ec '(^|[[:space:]])[[:alpha:]_][[:alnum:]_]*\(' cat_trace.txt)
PYTHON_COUNT=$(grep -Ec '(^|[[:space:]])[[:alpha:]_][[:alnum:]_]*\(' python_trace.txt)

cat > README.md <<EOF
# Практическая работа: системные вызовы Linux

## 1. Скрипт report.sh

Скрипт принимает каталог и уровень сообщений ERROR или WARN.
Он подсчитывает сообщения по модулям и сортирует результат по убыванию.
Дополнительно реализованы параметр --top N и проверка существования каталога.

Пример запуска:

\`\`\`bash
bash report.sh logs ERROR --top 3
\`\`\`

## 2. Программа hello.py

Программа читает файл и выводит его размер в байтах.

## 3. Анализ strace

Общее число системных вызовов Python: $TOTAL.

Общее число вызовов read(): $READ_COUNT.

Число системных вызовов cat: $CAT_COUNT.

Число системных вызовов Python: $PYTHON_COUNT.

Количество read() является общим для трассировки: оно не означает, что каждый вызов относится именно к sample.txt.

## 4. Вывод

Команда cat предназначена для вывода содержимого файла и выполняет сравнительно небольшой набор операций.

Python запускает интерпретатор, загружает библиотеки и модули, инициализирует среду выполнения и затем читает файл. Поэтому кроме операций чтения появляются дополнительные системные вызовы.

Разница в количестве системных вызовов показывает накладные расходы интерпретатора Python. Точные результаты зависят от версии системы, интерпретатора и окружения.
EOF

echo "README.md создан"
cat README.md
```
