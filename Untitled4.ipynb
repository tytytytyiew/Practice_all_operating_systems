```python
%%writefile ptree.py
import os
from pathlib import Path

PROC = Path("/proc")
MY_UID = os.getuid()
processes = {}

for entry in PROC.iterdir():
    if not entry.name.isdigit():
        continue

    pid = int(entry.name)

    try:
        status = (entry / "status").read_text()
        info = {}

        for line in status.splitlines():
            if ":" in line:
                key, value = line.split(":", 1)
                info[key] = value.strip()

        uid = int(info["Uid"].split()[0])

        if uid != MY_UID:
            continue

        stat = (entry / "stat").read_text()
        right = stat.rfind(")")
        fields = stat[right + 2:].split()

        ppid = int(fields[1])
        state = fields[0]
        rss_kb = int(info.get("VmRSS", "0 kB").split()[0])

        processes[pid] = {
            "ppid": ppid,
            "state": state,
            "rss": rss_kb,
            "name": info.get("Name", "?"),
        }

    except (FileNotFoundError, ProcessLookupError, PermissionError, ValueError, IndexError, KeyError):
        continue

children = {}

for pid, proc in processes.items():
    children.setdefault(proc["ppid"], []).append(pid)

for pid_list in children.values():
    pid_list.sort()

visited = set()

def print_tree(pid, prefix="", is_last=True):
    if pid in visited:
        return

    visited.add(pid)
    proc = processes[pid]
    connector = "└── " if is_last else "├── "

    print(
        f"{prefix}{connector if prefix else ''}"
        f"{proc['name']} PID={pid} "
        f"STATE={proc['state']} RSS={proc['rss']} kB"
    )

    next_prefix = prefix + ("    " if is_last else "│   ")

    child_list = children.get(pid, [])

    for index, child_pid in enumerate(child_list):
        print_tree(
            child_pid,
            next_prefix,
            index == len(child_list) - 1
        )

roots = sorted(
    pid for pid, proc in processes.items()
    if proc["ppid"] not in processes
)

for index, root in enumerate(roots):
    print_tree(root, "", index == len(roots) - 1)

for pid in sorted(processes):
    if pid not in visited:
        print_tree(pid)
```

```python
%%writefile runner.py
import os
import sys
import time

def main():
    if len(sys.argv) < 3:
        print("Использование: python3 runner.py N команда [аргументы...]")
        sys.exit(1)

    try:
        count = int(sys.argv[1])
        if count < 1:
            raise ValueError
    except ValueError:
        print("Ошибка: N должно быть положительным целым числом.")
        sys.exit(1)

    command = sys.argv[2:]
    children = {}

    for index in range(count):
        start = time.monotonic()
        pid = os.fork()

        if pid == 0:
            try:
                os.execvp(command[0], command)
            except OSError as error:
                print(f"Дочерний процесс: ошибка запуска: {error}", file=sys.stderr)
                os._exit(127)

        children[pid] = (index + 1, start)

    while children:
        pid, status = os.wait()
        end = time.monotonic()
        index, start = children.pop(pid)

        if os.WIFEXITED(status):
            code = os.WEXITSTATUS(status)
        elif os.WIFSIGNALED(status):
            code = f"signal {os.WTERMSIG(status)}"
        else:
            code = "unknown"

        print(
            f"Процесс #{index}: PID={pid}, "
            f"код завершения={code}, "
            f"время={end - start:.3f} сек."
        )

if __name__ == "__main__":
    main()
```

```python
%%writefile zombie.py
import os
import time

pid = os.fork()

if pid == 0:
    print(f"Дочерний процесс PID={os.getpid()} завершится.", flush=True)
    os._exit(7)

print(f"Родитель PID={os.getpid()}, ребёнок PID={pid}", flush=True)
print("Родитель пока не вызывает wait().", flush=True)

time.sleep(20)

result, status = os.waitpid(pid, 0)
print(f"Родитель забрал статус ребёнка PID={result}.", flush=True)
```

```python
%%bash
python3 zombie.py > zombie_output.txt 2>&1 &
PARENT=$!

sleep 2

cat zombie_output.txt

CHILD=$(grep -oE 'ребёнок PID=[0-9]+' zombie_output.txt | grep -oE '[0-9]+$')

echo
echo "Состояние дочернего процесса:"
ps -o pid,ppid,state,stat,cmd -p "$CHILD"

echo
echo "Дерево процессов:"
ps -o pid,ppid,state,stat,cmd --forest -p "$PARENT","$CHILD"

wait "$PARENT"
cat zombie_output.txt
```

```python
%%writefile orphan.py
import os
import time

pid = os.fork()

if pid == 0:
    print(
        f"Ребёнок: PID={os.getpid()}, "
        f"исходный PPID={os.getppid()}",
        flush=True
    )

    time.sleep(3)

    print(
        f"Ребёнок после завершения родителя: "
        f"PID={os.getpid()}, новый PPID={os.getppid()}",
        flush=True
    )

    time.sleep(5)
    os._exit(0)

print(
    f"Родитель PID={os.getpid()} завершится, "
    f"оставив ребёнка PID={pid}",
    flush=True
)

os._exit(0)
```

```python
%%bash
python3 orphan.py > orphan_output.txt 2>&1 &
LAUNCHER=$!

sleep 1

echo "Вывод программы:"
cat orphan_output.txt

echo
echo "Процессы orphan.py:"
ps -eo pid,ppid,state,cmd | grep '[o]rphan.py' || true

sleep 4

echo
echo "Вывод после переподчинения:"
cat orphan_output.txt

echo
echo "Проверка процессов:"
ps -eo pid,ppid,state,cmd | grep '[o]rphan.py' || true

wait "$LAUNCHER" 2>/dev/null || true
```

```python
%%writefile cow.py
import os
import sys

def memory_info():
    info = {}

    with open("/proc/self/status") as file:
        for line in file:
            if line.startswith("VmRSS:"):
                info["RSS"] = line.split(":", 1)[1].strip()

    try:
        with open("/proc/self/smaps_rollup") as file:
            for line in file:
                if line.startswith(("Pss:", "Private_Dirty:", "Shared_Clean:")):
                    key, value = line.split(":", 1)
                    info[key] = value.strip()
    except OSError:
        pass

    return info

def show(label):
    info = memory_info()
    print(f"\n{label}", flush=True)

    for key, value in info.items():
        print(f"{key}: {value}", flush=True)

data = list(range(1_000_000))

print(f"PID родителя: {os.getpid()}", flush=True)
print(f"Размер списка: {len(data):,} элементов", flush=True)

pid = os.fork()

if pid == 0:
    show("Ребёнок ДО изменения списка")

    for i in range(len(data)):
        data[i] = data[i] + 1

    show("Ребёнок ПОСЛЕ изменения списка")
    os._exit(0)

os.waitpid(pid, 0)
print("\nРодитель дождался завершения ребёнка.", flush=True)
```

```python
%%writefile README.md
# Практическая работа: процессы Linux

## 1. ptree.py — дерево процессов

Программа читает данные из /proc без использования ps для построения дерева.
Для процессов текущего UID выводятся PID, состояние, RSS в килобайтах.
Связи дерева определяются по PID и PPID.

Запуск:

    python3 ptree.py

## 2. runner.py — создание дочерних процессов

Программа создаёт N процессов через os.fork().
Каждый ребёнок запускает переданную команду через os.execvp().
Родитель ожидает завершения процессов через os.wait() и выводит код
завершения и время от момента создания ребёнка до сбора его статуса.

Примеры:

    python3 runner.py 3 /bin/sleep 1
    python3 runner.py 3 /bin/true

## 3. Зомби и сирота

### Зомби

В zombie.py ребёнок завершается, а родитель временно не вызывает wait().
В этот период статус завершившегося ребёнка сохраняется в таблице процессов.
В выводе ps зомби обычно имеет состояние Z.

После вызова waitpid() родителем статус забирается и запись зомби исчезает.

### Сирота

В orphan.py родитель завершает работу раньше ребёнка.
Ребёнок продолжает выполняться и получает нового родителя.

Фактический вывод ps и результаты экспериментов находятся в файле
process_results.txt.

Усыновителем является процесс, которому Linux переподчинил сироту.
Это может быть PID 1 (init/systemd) либо подходящий subreaper.
В отчёте необходимо назвать процесс по фактическому PPID и имени,
полученным в среде Google Colab.

## 4. Copy-on-Write

В cow.py большой список создаётся до fork().
После fork() ребёнок измеряет память, изменяет элементы списка и измеряет
память повторно.

RSS показывает резидентную память процесса, включая общие страницы.
Поэтому RSS не обязательно значительно увеличивается после изменения списка.
Дополнительные показатели Pss и Private_Dirty помогают увидеть,
как меняется распределение общей и приватной памяти.

Механизм Copy-on-Write позволяет процессам совместно использовать страницы
памяти до их изменения. При записи Linux создаёт отдельные копии затронутых
страниц.

## Общий вывод

/proc позволяет получать сведения о процессах без использования ps.
fork() создаёт дочерний процесс, exec() заменяет его программу,
wait()/waitpid() позволяют родителю получить статус завершения.
Зомби ожидает сбора статуса родителем, сирота продолжает работу после
завершения родителя, а Copy-on-Write помогает экономить память.
Точные значения PID, PPID, RSS и времени зависят от среды выполнения.
```
