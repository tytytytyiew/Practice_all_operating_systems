```python
import pandas as pd
import matplotlib.pyplot as plt
import heapq
from collections import deque
from copy import deepcopy

set1 = [
    {"pid": "P1", "arrival": 0, "burst": 8, "priority": 2},
    {"pid": "P2", "arrival": 1, "burst": 4, "priority": 1},
    {"pid": "P3", "arrival": 2, "burst": 9, "priority": 3},
    {"pid": "P4", "arrival": 3, "burst": 5, "priority": 2},
]

set2 = [
    {"pid": "P1", "arrival": 0, "burst": 10, "priority": 3},
    {"pid": "P2", "arrival": 1, "burst": 2, "priority": 1},
    {"pid": "P3", "arrival": 2, "burst": 7, "priority": 4},
    {"pid": "P4", "arrival": 3, "burst": 4, "priority": 2},
    {"pid": "P5", "arrival": 5, "burst": 3, "priority": 2},
]

print("Set 1")
display(pd.DataFrame(set1))

print("Set 2")
display(pd.DataFrame(set2))
```

```python
def simulate(processes, algorithm, **params):
    if algorithm not in {"FCFS", "SJF", "SRTF", "Priority", "RR"}:
        raise ValueError("Неизвестный алгоритм")

    if algorithm == "RR" and params.get("quantum", 0) < 1:
        raise ValueError("Квант должен быть >= 1")

    aging_interval = params.get("aging", 0)
    jobs = {
        p["pid"]: {
            **p,
            "remaining": p["burst"],
            "finish": None,
            "first": None,
            "wait": 0,
            "ready_since": p["arrival"],
        }
        for p in processes
    }

    order = {p["pid"]: i for i, p in enumerate(processes)}
    ready = []
    intervals = []
    current = None
    quantum_left = params.get("quantum", 1)
    t = 0
    completed = 0
    n = len(jobs)

    def add_interval(start, end, pid):
        if intervals and intervals[-1][2] == pid and intervals[-1][1] == start:
            intervals[-1] = (intervals[-1][0], end, pid)
        else:
            intervals.append((start, end, pid))

    def add_arrivals(now):
        for p in processes:
            pid = p["pid"]
            if p["arrival"] == now:
                jobs[pid]["ready_since"] = now
                ready.append(pid)

    def priority_key(pid, now):
        job = jobs[pid]
        waited = job["wait"] + max(0, now - job["ready_since"])
        aging_steps = waited // aging_interval if aging_interval else 0
        return (
            job["priority"] - aging_steps,
            job["arrival"],
            order[pid],
        )

    while completed < n:
        add_arrivals(t)

        if algorithm == "RR":
            if current is None and ready:
                current = ready.pop(0)
                quantum_left = params["quantum"]
        elif algorithm == "FCFS":
            if current is None and ready:
                current = min(ready, key=lambda pid: (
                    jobs[pid]["arrival"], order[pid]
                ))
                ready.remove(current)
        elif algorithm == "SJF":
            if current is None and ready:
                current = min(ready, key=lambda pid: (
                    jobs[pid]["burst"], jobs[pid]["arrival"], order[pid]
                ))
                ready.remove(current)
        elif algorithm == "SRTF":
            candidates = ready + ([current] if current is not None else [])
            if candidates:
                chosen = min(candidates, key=lambda pid: (
                    jobs[pid]["remaining"],
                    jobs[pid]["arrival"],
                    order[pid]
                ))
                if chosen != current:
                    if current is not None:
                        ready.append(current)
                        jobs[current]["ready_since"] = t
                    ready.remove(chosen)
                    current = chosen
        elif algorithm == "Priority":
            candidates = ready + ([current] if current is not None else [])
            if candidates:
                chosen = min(candidates, key=lambda pid: priority_key(pid, t))
                if chosen != current:
                    if current is not None:
                        ready.append(current)
                        jobs[current]["ready_since"] = t
                    ready.remove(chosen)
                    current = chosen

        if current is None:
            add_interval(t, t + 1, "IDLE")
            t += 1
            continue

        job = jobs[current]
        if job["first"] is None:
            job["first"] = t

        job["wait"] += 0
        add_interval(t, t + 1, current)
        job["remaining"] -= 1
        t += 1

        if algorithm == "RR":
            quantum_left -= 1

        if job["remaining"] == 0:
            job["finish"] = t
            completed += 1
            current = None
        elif algorithm == "RR" and quantum_left == 0:
            add_arrivals(t)
            job["ready_since"] = t
            ready.append(current)
            current = None

        for pid in ready:
            jobs[pid]["wait"] += 1

    return intervals


def calculate_metrics(processes, intervals):
    results = []
    finish_times = {}
    first_times = {}
    executed = {p["pid"]: 0 for p in processes}

    for start, end, pid in intervals:
        if pid == "IDLE":
            continue
        executed[pid] += end - start
        finish_times[pid] = end
        first_times.setdefault(pid, start)

    for p in processes:
        pid = p["pid"]
        completion = finish_times[pid]
        turnaround = completion - p["arrival"]
        waiting = turnaround - p["burst"]
        response = first_times[pid] - p["arrival"]

        results.append({
            "PID": pid,
            "Ожидание": waiting,
            "Оборот": turnaround,
            "Отклик": response,
        })

    total_time = max((end for _, end, _ in intervals), default=0)
    busy_time = sum(
        end - start for start, end, pid in intervals if pid != "IDLE"
    )

    metrics = {
        "Среднее ожидание": sum(r["Ожидание"] for r in results) / len(results),
        "Средний оборот": sum(r["Оборот"] for r in results) / len(results),
        "Средний отклик": sum(r["Отклик"] for r in results) / len(results),
        "Загрузка CPU, %": 100 * busy_time / total_time if total_time else 0,
    }
    return results, metrics


def compare_algorithms(processes):
    rows = []
    intervals_by_algorithm = {}

    configs = [
        ("FCFS", {}),
        ("SJF", {}),
        ("SRTF", {}),
        ("Priority", {}),
        ("RR", {"quantum": 3}),
    ]

    for name, params in configs:
        intervals = simulate(processes, name, **params)
        _, metrics = calculate_metrics(processes, intervals)
        rows.append({"Алгоритм": name, **metrics})
        intervals_by_algorithm[name] = intervals

    return pd.DataFrame(rows), intervals_by_algorithm


table1, schedules1 = compare_algorithms(set1)
display(table1.round(2))

table2, schedules2 = compare_algorithms(set2)
display(table2.round(2))
```

```python
arrived = set()
```

```python
def add_arrivals(now):
    for p in processes:
        pid = p["pid"]
        if p["arrival"] == now and pid not in arrived:
            arrived.add(pid)
            jobs[pid]["ready_since"] = now
            ready.append(pid)
```

```python
def priority_key(pid, now):
    job = jobs[pid]
    executed = job["burst"] - job["remaining"]
    waited = max(0, now - job["arrival"] - executed)
    aging_steps = waited // aging_interval if aging_interval else 0
    return (
        job["priority"] - aging_steps,
        job["arrival"],
        order[pid],
    )
```

```python
rr_results = []

for quantum in range(1, 11):
    intervals = simulate(set2, "RR", quantum=quantum)
    _, metrics = calculate_metrics(set2, intervals)

    switches = sum(
        1 for i in range(1, len(intervals))
        if intervals[i][2] != intervals[i - 1][2]
    )

    rr_results.append({
        "Квант": quantum,
        "Среднее ожидание": metrics["Среднее ожидание"],
        "Переключения": switches,
    })

rr_df = pd.DataFrame(rr_results)
display(rr_df.round(2))

plt.figure(figsize=(10, 5))
plt.plot(
    rr_df["Квант"], rr_df["Среднее ожидание"],
    marker="o", label="Среднее ожидание"
)
plt.plot(
    rr_df["Квант"], rr_df["Переключения"],
    marker="s", label="Переключения"
)
plt.xticks(range(1, 11))
plt.xlabel("Квант времени")
plt.ylabel("Значение")
plt.title("Round Robin: влияние размера кванта")
plt.grid(True)
plt.legend()
plt.show()
```

```python
starvation_set = [
    {"pid": "LOW", "arrival": 0, "burst": 8, "priority": 10},
] + [
    {"pid": f"HIGH{i}", "arrival": i, "burst": 2, "priority": 1}
    for i in range(1, 9)
]

without_aging = simulate(starvation_set, "Priority")
with_aging = simulate(starvation_set, "Priority", aging=2)

def process_finish(intervals, pid):
    return max(end for start, end, name in intervals if name == pid)

print("LOW без aging:", process_finish(without_aging, "LOW"))
print("LOW с aging:", process_finish(with_aging, "LOW"))
```

```python
%%writefile README.md
# Практическая работа: алгоритмы планирования CPU

## Реализованные алгоритмы

- FCFS — первым поступил, первым обслуживается.
- SJF — выбирается процесс с кратчайшим временем выполнения без вытеснения.
- SRTF — выбирается процесс с наименьшим оставшимся временем.
- Priority — вытесняющее планирование по приоритету.
- RR — циклическое планирование с параметром quantum.

## Метрики

Среднее ожидание — среднее время в очереди.
Средний оборот — среднее время от поступления до завершения.
Средний отклик — среднее время до первого запуска.
Загрузка CPU — доля времени, когда процессор занят.

## Результаты

Ниже необходимо вставить таблицы, полученные при запуске ноутбука,
для set1 и set2, а также график RR для квантов от 1 до 10.

## GIL и выбор алгоритма

Малый квант RR повышает частоту переключений. Большой квант
уменьшает переключения, но может увеличить время ожидания отдельных
процессов.

Aging постепенно улучшает эффективный приоритет ожидающего процесса,
уменьшая риск голодания.

## Проверка

SRTF и RR с quantum=3 необходимо сверить с эталонными интервалами set1.
Для точной проверки требуется использовать данные и соглашение об очереди
из условия преподавателя.
```
